# Kategorikal

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../assets/train.csv")
df.head(5)
```

## Fitur kategorikal
- Embarked
- Pclass 

## Ambil fitur

```python
categorical_cols = ['Embarked', 'Pclass']
df_cat = df[categorical_cols]

df_cat.head()
```

## Ambil dua sampel

```python
c1 = df_cat.iloc[0]
c2 = df_cat.iloc[1]
```

## Simple Matching Distance

```python
categorical_distance = np.mean(c1 != c2)
print("Categorical Distance:", categorical_distance)
```
