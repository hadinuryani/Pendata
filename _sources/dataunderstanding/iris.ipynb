# Distence Iris

Menghitung jarak data numerik menggunakan dataset Iris.
Dataset Iris hanya memiliki atribut numerik sehingga cocok untuk perhitungan jarak Euclidean dan Manhattan
Fitur numerik:
- sepal length
- sepal width
- petal length
- petal width

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

iris = load_iris()
df_iris = pd.DataFrame(iris.data, columns=iris.feature_names)
df_iris.head()
```

## cek tipe data numeric 

```python
df_numeric = df_iris.select_dtypes(include=[np.number])
print(df_numeric.dtypes)
```

## Ambil dua sampel

```python
point1 = df_numeric.iloc[0]
point2 = df_numeric.iloc[1]
```

## Euclidean Distance

```python
euclidean_distance = np.sqrt(np.sum((point1 - point2)**2))
print("Jarak Euclidean:", euclidean_distance)
```

## Manhattan Distance

```python
manhattan_distance = np.sum(abs(point1 - point2))
print("Jarak Manhattan:", manhattan_distance)
```
