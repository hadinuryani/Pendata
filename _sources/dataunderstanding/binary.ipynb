# Binary

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../assets/train.csv")
df.head(5)
```

Fitur biner Titanic
- Survived
- Sex (setelah encoding)

```python
# ubah Sex ke binary
df_binary = df[['Survived', 'Sex']].copy()
df_binary['Sex'] = df_binary['Sex'].map({'male': 1, 'female': 0})

df_binary.head()
```

## Ambil dua sampel

```python
b1 = df_binary.iloc[0]
b2 = df_binary.iloc[1]
```

## Jaccard Distance

```python
from scipy.spatial.distance import jaccard

jaccard_distance = jaccard(b1, b2)
print("Jaccard Distance:", jaccard_distance)
```
