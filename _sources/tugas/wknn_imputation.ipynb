# WKNN (Weighted K-Nearest Neighbor)

WKNN adalah pengembangan dari algoritma **K-Nearest Neighbor (KNN)** yang memberikan bobot pada setiap tetangga berdasarkan jaraknya. Tetangga yang lebih dekat memiliki pengaruh lebih besar dibandingkan tetangga yang lebih jauh.

Rumus bobot:

$$
w = \frac{1}{d^2}
$$

Keterangan:

- $w$ = bobot tetangga
- $d$ = jarak antara data target dengan data tetangga

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor

data={
'IPK':[2,3,4,2,3,4,2],
'PO':[2000000,3000000,2000000,2000000,3000000,4000000,3000000],
'JML':[2,3,2,3,2,3,np.nan]
}

df=pd.DataFrame(data)

train=df.iloc[:6]
test=df.iloc[6:]

scalerX=MinMaxScaler()
scalerY=MinMaxScaler()

X_train=scalerX.fit_transform(train[['IPK','PO']])
X_test=scalerX.transform(test[['IPK','PO']])

y_train=scalerY.fit_transform(train[['JML']])

def weight(dist):
    return 1/(dist**2)

model=KNeighborsRegressor(n_neighbors=6,weights=weight)
model.fit(X_train,y_train)

pred=model.predict(X_test)
hasil=scalerY.inverse_transform(pred)

print("Prediksi Missing Value:",hasil[0][0])
```

```python
print("Prediksi Normalisasi:",pred[0][0])
print("Prediksi Skala Asli:",hasil[0][0])
```
