# Decimal Scaling

Rumus:

$$
x' = \frac{x}{10^j}
$$

Dimana:

- $x$ = nilai asli  
- $j$ = jumlah digit dari nilai terbesar

```python
import pandas as pd
import numpy as np

data={
'IPK':[2,3,4,2,3,4],
'PO':[2000000,3000000,2000000,2000000,3000000,4000000],
'JML':[2,3,2,3,2,3]
}

df=pd.DataFrame(data)

df_scaled=df.copy()

for col in df.columns:
    max_val=df[col].abs().max()
    j=len(str(int(max_val)))
    df_scaled[col]=df[col]/(10**j)

df_scaled
```
