# Min-Max Normalization

### Rumus 

$$
x' = \frac{x - x_{min}}{x_{max} - x_{min}}
$$

Metode ini mengubah nilai data ke dalam rentang **0–1**.

Keterangan:

- $x'$ = nilai hasil normalisasi  
- $x$ = nilai asli  
- $x_{min}$ = nilai minimum pada dataset  
- $x_{max}$ = nilai maksimum pada dataset

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data={
'IPK':[2,3,4,2,3,4],
'PO':[2000000,3000000,2000000,2000000,3000000,4000000],
'JML':[2,3,2,3,2,3]
}

df=pd.DataFrame(data)

scaler=MinMaxScaler()
norm=scaler.fit_transform(df)

pd.DataFrame(norm,columns=df.columns)
```
