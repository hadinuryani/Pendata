# Z-Score Normalization

Rumus:

$$
z = \frac{x - \mu}{\sigma}
$$

Metode ini menstandarkan data sehingga memiliki:

- **Mean = 0**
- **Standar deviasi = 1**

Keterangan:

- $z$ = nilai hasil normalisasi  
- $x$ = nilai asli data  
- $\mu$ = rata-rata (mean) dari atribut  
- $\sigma$ = standar deviasi dari atribut

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

data={
'IPK':[2,3,4,2,3,4],
'PO':[2000000,3000000,2000000,2000000,3000000,4000000],
'JML':[2,3,2,3,2,3]
}

df=pd.DataFrame(data)

scaler=StandardScaler()
z=scaler.fit_transform(df)

pd.DataFrame(z,columns=df.columns)
```
