## Missing Values

Missing value adalah kondisi ketika suatu atribut pada dataset tidak memiliki nilai.

Beberapa penyebab terjadinya missing value antara lain:

- Kesalahan dalam proses pengumpulan data
- Data tidak diisi oleh responden
- Kesalahan pada sistem penyimpanan data

Beberapa metode yang dapat digunakan untuk menangani missing value antara lain:

1. Menghapus data yang memiliki nilai kosong
2. Mengisi nilai dengan mean atau median
3. Menggunakan metode imputasi seperti KNN atau WKNN

---

## Normalisasi Data

Normalisasi merupakan proses mengubah atau menskalakan nilai data ke dalam rentang tertentu agar semua atribut memiliki skala yang sama atau sebanding.

Tujuan normalisasi adalah agar tidak ada fitur yang terlalu dominan karena perbedaan skala ketika data dianalisis atau digunakan dalam algoritma machine learning.

Metode normalisasi yang sering digunakan antara lain:

1. Min-Max Normalization
2. Z-Score Normalization
3. Decimal Scaling
