# Min-Max Normalization

---

## Pengertian

**Min-Max Normalization** merupakan salah satu metode normalisasi data yang digunakan untuk mengubah skala nilai suatu atribut numerik ke dalam rentang tertentu. Biasanya rentang yang digunakan adalah **0 sampai 1**, namun dalam beberapa kasus juga dapat menggunakan rentang lain seperti -1 sampai 1.

Tujuan utama dari normalisasi ini adalah agar semua atribut memiliki **skala yang sebanding**. Dalam dataset sering ditemukan atribut yang memiliki rentang nilai sangat berbeda — misalnya satu atribut bernilai antara 1–5 sedangkan atribut lain bernilai ratusan ribu atau jutaan. Jika data tersebut langsung digunakan dalam perhitungan, maka atribut dengan nilai yang lebih besar dapat mendominasi hasil analisis.

## Rumus

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$

Untuk rentang kustom $[new\_min,\ new\_max]$:

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}} \times (new\_max - new\_min) + new\_min$$

**Keterangan:**
- $x$ : nilai asli dari data
- $x'$ : nilai baru setelah dinormalisasi
- $x_{min}$ : nilai minimum atribut pada data asli
- $x_{max}$ : nilai maksimum atribut pada data asli

## Kelebihan dan Kekurangan

| Kelebihan | Kekurangan |
|-----------|------------|
| Hasil selalu berada di rentang [0, 1] | Sangat sensitif terhadap outlier |
| Mempertahankan distribusi asli data | Jika ada data baru di luar range, perlu recalculate |
| Mudah diinterpretasikan | Nilai min/max harus diketahui |
| Cocok untuk algoritma distance-based (KNN, WKNN) | |

## Contoh Perhitungan Manual

### Kolom IPK
- $x_{min} = 2$, $x_{max} = 4$

**Objek ke-1** (IPK = 2):
$$x' = \frac{2 - 2}{4 - 2} = \frac{0}{2} = 0$$

**Objek ke-2** (IPK = 3):
$$x' = \frac{3 - 2}{4 - 2} = \frac{1}{2} = 0.5$$

**Objek ke-3** (IPK = 4):
$$x' = \frac{4 - 2}{4 - 2} = \frac{2}{2} = 1$$

### Kolom PO
- $x_{min} = 2.000.000$, $x_{max} = 4.000.000$

**Objek ke-2** (PO = 3.000.000):
$$x' = \frac{3.000.000 - 2.000.000}{4.000.000 - 2.000.000} = \frac{1.000.000}{2.000.000} = 0.5$$

### Kolom JML
- $x_{min} = 2$, $x_{max} = 3$

**Objek ke-2** (JML = 3):
$$x' = \frac{3 - 2}{3 - 2} = \frac{1}{1} = 1$$

## Implementasi Python — Fungsi Manual

```python
import pandas as pd
import numpy as np

# Dataset
data = {
    'No':  [1, 2, 3, 4, 5, 6],
    'IPK': [2, 3, 4, 2, 3, 4],
    'PO':  [2000000, 3000000, 2000000, 2000000, 3000000, 4000000],
    'JML': [2, 3, 2, 3, 2, 3]
}
df = pd.DataFrame(data).set_index('No')

print("=== Data Sebelum Normalisasi ===")
print(df)
```

```python
# =========================================
# Fungsi Min-Max Normalization Manual
# =========================================
def minmax_normalization(df, kolom):
    """
    Melakukan Min-Max Normalization pada kolom yang dipilih.
    
    Parameters:
        df    : DataFrame
        kolom : list kolom yang akan dinormalisasi
    
    Returns:
        DataFrame hasil normalisasi
    """
    df_norm = df.copy()
    for col in kolom:
        x_min = df[col].min()
        x_max = df[col].max()
        df_norm[col] = (df[col] - x_min) / (x_max - x_min)
        print(f"Kolom {col}: min={x_min}, max={x_max}")
    return df_norm

kolom_target = ['IPK', 'PO', 'JML']
df_manual = minmax_normalization(df, kolom_target)

print("\n=== Hasil Min-Max Normalization (Fungsi Manual) ===")
print(df_manual.round(4))
```

```python
# Detail perhitungan per objek untuk kolom IPK
print("=== Detail Perhitungan Kolom IPK ===")
x_min = df['IPK'].min()
x_max = df['IPK'].max()
print(f"x_min = {x_min}, x_max = {x_max}\n")

for idx, val in df['IPK'].items():
    norm = (val - x_min) / (x_max - x_min)
    print(f"Objek {idx}: ({val} - {x_min}) / ({x_max} - {x_min}) = {norm:.4f}")
```

## Implementasi Python — Menggunakan sklearn

```python
from sklearn.preprocessing import MinMaxScaler

# =========================================
# MinMaxScaler dari sklearn
# =========================================
scaler = MinMaxScaler(feature_range=(0, 1))  # rentang default [0, 1]

df_sklearn = df.copy()
df_sklearn[kolom_target] = scaler.fit_transform(df[kolom_target])

print("=== Hasil Min-Max Normalization (sklearn MinMaxScaler) ===")
print(df_sklearn.round(4))

print("\nNilai min per kolom (dari scaler):", scaler.data_min_)
print("Nilai max per kolom (dari scaler):", scaler.data_max_)
```

```python
# Verifikasi: hasil manual vs sklearn harus sama
print("=== Verifikasi: Manual vs sklearn ===")
print(f"Hasil identik: {df_manual.round(6).equals(df_sklearn.round(6))}")

print("\nPerbandingan kolom IPK:")
print(pd.DataFrame({
    'Original' : df['IPK'],
    'Manual'   : df_manual['IPK'].round(4),
    'sklearn'  : df_sklearn['IPK'].round(4)
}))
```

## Kesimpulan

- Min-Max Normalization mengubah nilai data ke rentang **[0, 1]** dengan rumus $x' = (x - x_{min}) / (x_{max} - x_{min})$
- Metode ini **mempertahankan distribusi** asli data dan mudah diinterpretasikan
- Sangat cocok digunakan sebelum algoritma berbasis jarak seperti **KNN dan WKNN**
- Kelemahan utama: sensitif terhadap **outlier** — satu nilai ekstrem bisa menekan semua nilai lain ke rentang yang sempit
