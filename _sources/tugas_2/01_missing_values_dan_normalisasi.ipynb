## 1. Missing Values

### Pengertian

**Missing values** (nilai yang hilang) adalah kondisi di mana suatu atribut pada baris tertentu dalam dataset tidak memiliki nilai. Hal ini sangat umum terjadi dalam data nyata dan dapat disebabkan oleh berbagai faktor.

### Penyebab Missing Values

- **Kesalahan input data** — data tidak diisi saat pengumpulan
- **Kerusakan data** — data rusak atau hilang selama proses transfer
- **Tidak relevan** — responden tidak mengisi karena pertanyaan tidak berlaku
- **Kesalahan sistem** — error pada perangkat atau aplikasi pengumpul data

### Dampak Missing Values

Jika missing values dibiarkan, maka dapat menyebabkan:
- Hasil analisis yang **bias** dan tidak akurat
- Banyak algoritma machine learning **tidak dapat berjalan** dengan data kosong
- Pengurangan ukuran dataset jika baris kosong dihapus

### Cara Menangani Missing Values

| Metode | Deskripsi | Kelebihan | Kekurangan |
|--------|-----------|-----------|------------|
| **Deletion** | Menghapus baris/kolom yang memiliki missing value | Sederhana | Kehilangan banyak data |
| **Mean/Median/Mode Imputation** | Mengisi dengan rata-rata, median, atau modus | Mudah diterapkan | Tidak memperhitungkan hubungan antar fitur |
| **KNN Imputation** | Mengisi berdasarkan K tetangga terdekat | Mempertimbangkan hubungan antar data | Komputasi lebih berat |
| **WKNN Imputation** | KNN dengan pembobotan jarak | Lebih akurat dari KNN biasa | Komputasi lebih berat |
| **Regression Imputation** | Mengisi dengan hasil prediksi regresi | Akurat untuk data linear | Asumsi linearitas |

## 2. Normalisasi Data

### Pengertian

**Normalisasi data** adalah proses mengubah skala nilai-nilai dalam suatu dataset agar berada dalam rentang tertentu atau memiliki distribusi tertentu. Proses ini merupakan bagian penting dari tahap **preprocessing** data sebelum digunakan dalam analisis atau pemodelan machine learning.

### Mengapa Normalisasi Penting?

Dalam sebuah dataset, atribut yang berbeda seringkali memiliki skala nilai yang sangat berbeda. Misalnya:

| Atribut | Rentang Nilai |
|---------|---------------|
| IPK     | 0 – 4         |
| PO      | 2.000.000 – 4.000.000 |
| JML     | 1 – 5         |

Jika data langsung digunakan tanpa normalisasi, atribut dengan nilai besar (seperti PO) akan **mendominasi** perhitungan jarak pada algoritma seperti KNN dan WKNN, sehingga atribut lain menjadi tidak berpengaruh secara signifikan.

### Macam-Macam Normalisasi Data

Terdapat tiga metode normalisasi yang umum digunakan:

#### a) Min-Max Normalization
Mengubah nilai ke dalam rentang **[0, 1]**.

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$

-  Mempertahankan distribusi asli data
-  Sangat sensitif terhadap outlier

#### b) Z-Score Normalization (Standardisasi)
Mengubah data sehingga memiliki **mean = 0** dan **standar deviasi = 1**.

$$z = \frac{x - \mu}{\sigma}$$

-  Lebih tahan terhadap outlier
-  Cocok untuk algoritma yang mengasumsikan distribusi normal
-  Tidak memiliki batas atas/bawah yang pasti

#### c) Decimal Scaling
Menggeser titik desimal sehingga nilai absolut terbesar menjadi kurang dari 1.

$$v' = \frac{v}{10^j}$$

- Sangat sederhana
- Cocok untuk data berskala besar
- Hasil normalisasi tidak selalu berada di [0, 1]

### Perbandingan Ketiga Metode

| Aspek | Min-Max | Z-Score | Decimal Scaling |
|-------|---------|---------|------------------|
| Rentang hasil | [0, 1] | ~[-3, 3] | (-1, 1) |
| Sensitif outlier | Ya | Tidak | Tidak |
| Kompleksitas | Rendah | Sedang | Sangat rendah |
| Menjaga distribusi | Ya | Tidak (mengubah ke normal) | Ya |

```python
import pandas as pd
import numpy as np

# Dataset contoh
data = {
    'No':  [1, 2, 3, 4, 5, 6, 7],
    'IPK': [2, 3, 4, 2, 3, 4, 2],
    'PO':  [2000000, 3000000, 2000000, 2000000, 3000000, 4000000, 3000000],
    'JML': [2, 3, 2, 3, 2, 3, None]   # Objek ke-7 = missing value
}

df = pd.DataFrame(data).set_index('No')

print("=== Dataset dengan Missing Value ===")
print(df)
print(f"\nJumlah missing values per kolom:")
print(df.isnull().sum())
```

```python
# Ilustrasi perbedaan skala antar atribut
df_train = df.dropna()

print("=== Statistik Deskriptif Dataset (tanpa missing) ===")
print(df_train.describe().round(2))

print("\n  Perhatikan perbedaan skala:")
print(f"   Rentang IPK : {df_train['IPK'].min()} – {df_train['IPK'].max()}")
print(f"   Rentang PO  : {df_train['PO'].min():,} – {df_train['PO'].max():,}")
print(f"   Rentang JML : {df_train['JML'].min()} – {df_train['JML'].max()}")
print("\n→ Perbedaan skala ini yang membuat normalisasi menjadi penting.")
```

## Kesimpulan

- **Missing values** harus ditangani sebelum data dianalisis agar hasil tidak bias.
- **Normalisasi data** diperlukan agar setiap atribut memiliki kontribusi yang setara dalam perhitungan, terutama pada algoritma berbasis jarak seperti KNN dan WKNN.
- Tiga metode normalisasi yang umum: **Min-Max**, **Z-Score**, dan **Decimal Scaling**, masing-masing cocok untuk kondisi data yang berbeda.
